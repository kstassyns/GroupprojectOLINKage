```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

np.random.seed(42)
os.makedirs("results", exist_ok=True)
```

```python
meta = pd.read_csv("sample_metadata.csv")
raw = pd.read_csv("proteins_npx_raw.csv")
anno = pd.read_csv("protein_annotation.csv")

print(meta.shape, raw.shape, anno.shape)
print(meta.info())
print(anno.head())
print(raw.iloc[:, :6].head())

assert meta["sample_kit_id"].is_unique
assert raw["sample_kit_id"].is_unique
assert anno["protein"].is_unique
assert set(meta["sample_kit_id"]) == set(raw["sample_kit_id"])
assert set(anno["protein"]) == set(raw.columns[1:])

# Metadata exact in dezelfde volgorde zetten als raw
meta = meta.set_index("sample_kit_id").loc[raw["sample_kit_id"]].reset_index()
assert meta["sample_kit_id"].equals(raw["sample_kit_id"])

X = raw.set_index("sample_kit_id").copy()
X = X.apply(pd.to_numeric, errors="coerce")
X.index.name = "sample_kit_id"
anno = anno.set_index("protein").loc[X.columns].reset_index()
assert X.select_dtypes(include=np.number).shape == X.shape
```
